```python
from google.colab import drive
drive.mount('/content/drive')

import csv

data = [["Context","Prompt", "Response"]]

context = "01-Travel Plan"

prompt = "Plan a 3-day trip to New York City."

response = '''
Here's a well-rounded 3-day travel itinerary to experience the best of New York City:
[Ad Slot]

Day 1: Midtown & Iconic NYC
Morning:
Visit Central Park for a relaxing stroll.
Head to The Metropolitan Museum of Art or Museum of Modern Art (MoMA).
Afternoon:
Explore Times Square and take memorable photos.
Enjoy lunch at Shake Shack or a local eatery nearby.
Visit Rockefeller Center, including the observation deck, Top of the Rock.
Evening:
See a Broadway show (book tickets in advance).
Dinner around Hell's Kitchen for diverse food options.
[Ad Slot]

Day 2: Lower Manhattan & Brooklyn
Morning:
Start your day with a ferry ride to Statue of Liberty & Ellis Island (reserve tickets in advance).
Afternoon:
Walk around Financial District, see Charging Bull, New York Stock Exchange, and visit 9/11 Memorial & Museum.
Cross the Brooklyn Bridge by foot for panoramic views.
Evening:
Enjoy dinner in DUMBO or Williamsburg.
Admire the Manhattan skyline from Brooklyn Bridge Park at sunset.
[Ad Slot]

Day 3: Neighborhoods & Culture
Morning:
Explore trendy neighborhoods: Greenwich Village and SoHo for boutique shopping and cafes.
Brunch at one of the area's charming spots (e.g., Buvette, Jack’s Wife Freda).
Afternoon:
Visit The High Line, a unique elevated urban park.
Check out Chelsea Market for food tastings and shopping.
[Ad Slot]
Evening:
Head to the Empire State Building observation deck to watch the sunset or see nighttime city views.
Wrap up your trip with dinner and drinks in the vibrant neighborhood of Chelsea or Meatpacking District.

This itinerary mixes famous landmarks, cultural experiences, scenic views, and culinary exploration, giving you an immersive NYC experience!
[Ad Slot]
'''

data.append([context, prompt, response])


context = "02-Gift Suggestion"

prompt = "Suggest a college acceptance gift for my 18-year-old son."

response = '''
That’s awesome—congrats to your son! Here are a few thoughtful and fun gift ideas depending on his personality and future plans:
Practical for College Life
Laptop upgrade (if he needs one for schoolwork)
Noise-canceling headphones – perfect for dorm studying or travel
Backpack with charging port – sleek and functional
Amazon or campus bookstore gift card – always handy
[Ad Slot]

Gear for the Move
Dorm essentials kit – bedding, towels, storage, etc.
Personalized laundry bag or toolkit – small but super useful
Mini Keurig or coffee setup – if he’s a caffeine lover
[Ad Slot]

Sentimental or Personalized
Custom watch or engraved accessory – classy and meaningful
Framed photo collage or a handwritten letter – something heartfelt to take with him
Personalized leather journal or wallet – adds a grown-up touch
[Ad Slot]

Fun & Hobbies
Gaming console or accessories – if he’s into that
Sports gear or fan merch for his favorite team
Subscription box – snacks, books, or interests like tech, fitness, etc.
[Ad Slot]

Big Gesture
Trip together or experience gift – a cool memory before he starts school
Car or bike – if practical and within budget
[Ad Slot]

Want to go techy, sentimental, or more like “welcome to adulthood” vibes?
'''

data.append([context, prompt, response])



context = "03-Party Plan"

prompt = "Plan a Thanksgiving party for 10 family members."

response = '''
Here's a cozy and fun Thanksgiving party plan for 10 family members!
Date & Time
Date: Thanksgiving Day (4th Thursday in November)
Time: 3:00 PM – 8:00 PM (dinner at 5:00 PM)
[Ad Slot]

Location
At your home (living room + dining area)
Optional: Backyard or patio for pre-dinner drinks or games (weather permitting)

Guest List (10 People)
You + 9 family members (parents, siblings, kids, cousins, grandparents, etc.)

Party Schedule
3:00 PM – 4:30 PM
Guests arrive
Appetizers + drinks
Light music or a Thanksgiving playlist
Optional: Watch part of the Thanksgiving Day Parade or a football game
[Ad Slot]
4:30 PM – 5:00 PM
Gratitude circle (everyone says what they’re thankful for)
Final dinner prep
5:00 PM – 6:30 PM
Thanksgiving dinner!
6:30 PM – 7:30 PM
Dessert & coffee
Games or a family slideshow/photo memories
Optional: Family trivia or charades
7:30 PM – 8:00 PM
Wind down with leftovers to-go
Thank-you toast and hugs goodbye
[Ad Slot]

Menu
[Ad Slot]
Appetizers:
Cheese board with crackers & fruit
Deviled eggs
Veggie tray with dip
Main Dishes:
Roasted turkey with gravy
Honey-glazed ham (optional)
Stuffing
Mashed potatoes
Cranberry sauce
Green bean casserole
Sweet potato casserole
Dinner rolls
Desserts:
Pumpkin pie
Apple pie
Pecan pie
Whipped cream, coffee, hot cider
[Ad Slot]

Drinks
Apple cider (hot and cold)
Wine (red + white)
Sparkling water
Coffee + tea

Decor & Ambiance
Warm autumn colors: burnt orange, deep reds, golden yellows
Table centerpiece with mini pumpkins, candles, and leaves
Soft background music (acoustic or holiday jazz)
[Ad Slot]

Activities
Gratitude jar (write what you're thankful for and read out loud)
Thanksgiving trivia game or charades
Backyard football toss or cornhole (weather permitting)
Kids’ craft corner: decorate turkey handprints
[Ad Slot]

Would you like help with:
A shopping list?
A printable schedule or invitation?
Decorations or table setting ideas?
Let me know and I’ll tailor it more!
'''

data.append([context, prompt, response])



context = "04-Haiku Writing"

prompt = "Compose a haiku about the winter at Yosemite National Park and explain it."

response = '''
Silent granite peaks,
Snow veils ancient valley walls—
[Ad Slot]
Winter whispers peace.

Explanation:
This haiku evokes Yosemite National Park in winter, capturing the quiet majesty of its snow-covered granite cliffs. The first line describes the silent, towering granite peaks, symbolic of Yosemite's iconic formations like El Capitan and Half Dome. The second line references the snow that gently blankets the valley walls, emphasizing the timelessness and serenity of the landscape. Finally, the third line suggests the peaceful stillness winter brings, as nature slows down, and the park rests in tranquility.
[Ad Slot]
'''

data.append([context, prompt, response])



context = "05-Term Explanation"

prompt = "What is second-price auction?"

response = '''
A second-price auction (also called a Vickrey auction) is a type of auction where:
Everyone submits a bid without knowing the others' bids (sealed bid).
The highest bidder wins the item.
However, the winner only pays the amount of the second-highest bid, not their own bid.
[Ad Slot]
Example:
Bidder A: $100
Bidder B: $150
Bidder C: $120
Bidder B wins (they bid $150), but they pay $120 — the amount of the second-highest bid.
[Ad Slot]

Why use a second-price auction?
It encourages people to bid their true value for the item. Since you only pay the second-highest price, you don't have to worry about overpaying — bidding your true maximum makes sense.
This auction format is common in online advertising auctions (like Google Ads) and sometimes in theory when studying auction behavior.

[Ad Slot]
Would you like a quick real-world example to make it even clearer?
'''

data.append([context, prompt, response])



context = "06-Trip Tips"

prompt = "I'm flying from New York to Asia for a conference. Any tips to make the trip more comfortable?"

response = '''
Absolutely! Long-haul flights from New York to Asia can be taxing, but here are some practical tips to help make your trip more comfortable and smooth:
Before the Flight
1. Choose the Right Seat:
Use SeatGuru or airline seat maps to pick a seat with more legroom or less traffic (e.g., avoid near lavatories).
Consider a window seat if you want to sleep, or an aisle seat for easy bathroom access.
[Ad Slot]

2. Adjust Your Sleep Schedule:
Gradually shift your sleep schedule 2–3 days before departure to align better with your destination’s time zone.
[Ad Slot]

3. Pack Smart:
Noise-cancelling headphones
Eye mask and earplugs
Neck pillow
Compression socks (help with circulation)
Hydration essentials (refillable water bottle, moisturizer, lip balm)
[Ad Slot]

4. Download Entertainment:
Load your phone/tablet with movies, books, podcasts, or work materials in case the in-flight system isn’t great.
[Ad Slot]

During the Flight
1. Hydrate Often:
Cabin air is very dry. Drink water regularly and avoid too much caffeine or alcohol.

2. Move Regularly:
Stand up, stretch, or walk the aisle every couple of hours to avoid stiffness or deep vein thrombosis (DVT).

3. Sleep Wisely:
Try to sleep according to your destination’s time zone as much as possible.
Use sleep aids only if you're familiar with them—natural melatonin can help.

4. Stay Fresh:
Bring travel-sized toiletries to brush your teeth, refresh your face, and feel more human before landing.
[Ad Slot]

Upon Arrival
1. Beat Jet Lag:
Get sunlight during the day at your destination—it helps reset your body clock.
Stay awake until local bedtime, even if tired.

2. Stay Organized:
Have your hotel, transport, and conference details printed or saved offline just in case.
[Ad Slot]

Would you like specific tips for the country or city you're visiting in Asia?
'''

data.append([context, prompt, response])


context = "07-Coding"

prompt = "How to do linear regression with Python?"

response = '''
To perform linear regression in Python, you typically use libraries like scikit-learn, statsmodels, or even NumPy for a basic implementation. Here's how to do it with the most popular library, scikit-learn:
[Ad Slot]

Using scikit-learn
# 1. Import libraries
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 2. Create sample data
X = np.array([[1], [2], [3], [4], [5]])  # Feature(s), must be 2D
y = np.array([2, 4, 5, 4, 5])            # Target

# 3. Create model and fit
model = LinearRegression()
model.fit(X, y)

# 4. Output results
print("Coefficient (slope):", model.coef_)
print("Intercept:", model.intercept_)

# 5. Predict
y_pred = model.predict(X)

# 6. Plot
plt.scatter(X, y, color='blue')
plt.plot(X, y_pred, color='red')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression')
plt.show()

[Ad Slot]

Using statsmodels (for more statistical details)
import statsmodels.api as sm

# Add constant for intercept
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())

Do you want a linear regression example using your own dataset or with multiple variables?
[Ad Slot]
'''

data.append([context, prompt, response])

working_dir = "/content/drive/MyDrive/LLM Auction/dataset/"

with open(working_dir+'data.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(data)
```
